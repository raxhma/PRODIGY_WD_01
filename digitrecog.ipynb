```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

# =========================
# Load MNIST dataset
# =========================
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# =========================
# Normalize data
# =========================
x_train = tf.keras.utils.normalize(x_train, axis=1)
x_test = tf.keras.utils.normalize(x_test, axis=1)

# =========================
# Build model
# =========================
model = tf.keras.models.Sequential([
    tf.keras.Input(shape=(28, 28)), # Explicitly define the input layer
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

# =========================
# Compile model
# =========================
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# =========================
# Train model
# =========================
model.fit(x_train, y_train, epochs=3)

# =========================
# Save model
# =========================
model.save('handwritten.keras')

# =========================
# Load model
# =========================
model = tf.keras.models.load_model('handwritten.keras')

# =========================
# Predict custom images
# =========================
image_number = 1

while os.path.isfile(f"digits/img_{image_number}.jpg"):
    img = cv2.imread(f"digits/img_{image_number}.jpg")[:, :, 0]
    img = np.invert(np.array([img]))

    prediction = model.predict(img)

    print(f"This digit is probably a {np.argmax(prediction)}")

    plt.imshow(img[0], cmap=plt.cm.binary)
    plt.show()

    image_number += 1
```

```python
image_to_predict = 10 # @param {type:"integer"}

image_path = f"digits/img_{image_to_predict}.jpg"

if os.path.isfile(image_path):
    img = cv2.imread(image_path)[:, :, 0]
    # Invert colors to match MNIST format (white digit on black background)
    img = np.invert(np.array([img]))

    # Normalize the custom image in the same way as the training data
    # tf.keras.utils.normalize expects float input
    img_normalized = tf.keras.utils.normalize(img.astype('float32'), axis=1)

    prediction = model.predict(img_normalized)

    print(f"This digit is probably a {np.argmax(prediction)}")

    plt.imshow(img[0], cmap=plt.cm.binary)
    plt.title(f"Predicted: {np.argmax(prediction)}")
    plt.show()
else:
    print(f"Error: Image file '{image_path}' not found. Please ensure the image exists and the number is correct.")
```

```python
# Identify misclassifications from the confusion matrix
misclassifications = []

for i in range(cm.shape[0]): # Iterate over true labels
    for j in range(cm.shape[1]): # Iterate over predicted labels
        if i != j and cm[i, j] > 0: # If it's an off-diagonal element and not zero
            misclassifications.append({
                'true_label': i,
                'predicted_label': j,
                'count': cm[i, j]
            })

# Sort misclassifications by count in descending order
misclassifications.sort(key=lambda x: x['count'], reverse=True)

print("Top misclassifications:")
for mc in misclassifications[:10]: # Display top 10 misclassifications
    print(f"  True digit: {mc['true_label']}, Predicted as: {mc['predicted_label']}, Count: {mc['count']}")

# Identify which true digits are most often misclassified
# This involves summing misclassification counts across rows (excluding the diagonal)

misclassified_counts_by_true_digit = np.sum(cm, axis=1) - np.diag(cm)

print("\nDigits most frequently misclassified (true label):")
for i, count in enumerate(misclassified_counts_by_true_digit):
    if count > 0:
        print(f"  Digit {i}: {int(count)} times")

# Identify which predicted digits are most often wrongly predicted
# This involves summing misclassification counts across columns (excluding the diagonal)

misclassified_counts_by_predicted_digit = np.sum(cm, axis=0) - np.diag(cm)

print("\nDigits most frequently predicted incorrectly (predicted label):")
for i, count in enumerate(misclassified_counts_by_predicted_digit):
    if count > 0:
        print(f"  Digit {i}: {int(count)} times")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Get predictions for the entire test set
y_pred_probabilities = model.predict(x_test)
y_pred_labels = np.argmax(y_pred_probabilities, axis=1)

# Compute the confusion matrix
cm = confusion_matrix(y_test, y_pred_labels)

# Plot the confusion matrix
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

num_images_to_display = 5 # You can change this number

print(f"Displaying predictions for the first {num_images_to_display} test images:")

for i in range(num_images_to_display):
    # Reshape the image for prediction (add batch dimension)
    img_for_prediction = np.expand_dims(x_test[i], axis=0)

    # Make prediction
    prediction = model.predict(img_for_prediction)
    predicted_label = np.argmax(prediction)
    true_label = y_test[i]

    # Display the image
    plt.imshow(x_test[i], cmap=plt.cm.binary)
    plt.title(f"True: {true_label}, Predicted: {predicted_label}")
    plt.axis('off') # Hide axes
    plt.show()
```

```python
loss, accuracy = model.evaluate(x_test, y_test)
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
```
